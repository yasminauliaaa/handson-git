```python
import pandas as pd
from datetime import datetime
```

```python
# STEP 1 — LOAD DATA & CEK AWAL
```

```python
df = pd.read_csv('pertamina_oil_gas_raw.csv')

print("===== CEK DATA AWAL =====")
print(f"Jumlah baris & kolom : {df.shape}")
print(f"Duplikat             : {df.duplicated().sum()} baris")
print(f"Missing values       : {df.isnull().sum().sum()}")
print(f"\nContoh format tanggal:")
print(df['date'].sample(5, random_state=1).tolist())
```

```python
# STEP 2 — HAPUS DUPLIKAT
```

```python
df = df.drop_duplicates()

print(f"\n✅ STEP 2 — Duplikat dihapus")
print(f"   Sisa baris: {len(df)}")
```

```python
# STEP 3 — CEK MISSING VALUES PER KOLOM
```

```python
print(f"\n===== CEK MISSING VALUES =====")
print(df.isnull().sum())
print(f"\nTotal missing: {df.isnull().sum().sum()} → tidak ada yang perlu diisi ✅")
```

```python
# STEP 4 — STANDARDISASI FORMAT TANGGAL
# 3 format: YYYY-MM-DD, DD/MM/YYYY, YYYY/MM/DD
```

```python
def parse_tanggal(tgl):
    for fmt in ['%Y-%m-%d', '%d/%m/%Y', '%Y/%m/%d']:
        try:
            return datetime.strptime(str(tgl), fmt)
        except:
            continue
    return None

df['date']    = pd.to_datetime(df['date'].apply(parse_tanggal))
df['year']    = df['date'].dt.year
df['month']   = df['date'].dt.month
df['quarter'] = df['date'].dt.to_period('Q').astype(str)

print(f"\n✅ STEP 4 — Tanggal distandarisasi")
print(f"   Range: {df['date'].min().date()} → {df['date'].max().date()}")
```

```python
# STEP 5 — SIMPAN HASIL
```

```python
df = df.sort_values('date').reset_index(drop=True)
df.to_csv('pertamina_oil_gas_clean.csv', index=False)

print(f"\n========== HASIL ==========")
print(f"Baris awal  : 512")
print(f"Baris akhir : {len(df)}")
print(f"Missing     : {df.isnull().sum().sum()}")
print(f"\n✅ Tersimpan: pertamina_oil_gas_clean.csv")
```
